```python
import pandas as pd
import numpy as np
#from pandas_profiling import ProfileReport
import seaborn as sns
import matplotlib.pyplot as plt

# Your analysis code here
```

```python
# Load the Excel file
file_path = r"D:\Study,(1).xlsx"  
sheet_name = 'Form Responses 1'  

# Load survey data
survey_data = pd.read_excel(file_path, sheet_name=sheet_name)

# Display the first few rows
print(survey_data.head())
```

```python
def check_missing_values(data):
    missing_summary = data.isnull().sum()
    print("\nMissing Values per Column:\n", missing_summary)
    return missing_summary
```

```python
def check_duplicates(data):
    duplicates = data.duplicated().sum()
    print(f"\nNumber of Duplicate Rows: {duplicates}")
    return duplicates
```

```python
def check_categorical_consistency(data, column_name, valid_values):
    invalid_values = data[~data[column_name].isin(valid_values)][column_name].unique()
    print(f"\nInvalid Values in '{column_name}': {invalid_values}")
    return invalid_values
```

```python
def check_logical_relationships(data):
    invalid_logic = data[
        (data['Familiarity_with_AI'] == 'Very familiar') & 
        (data['AI_Usage_Level'] == 'Not used')
    ]
    print(f"\nRows Violating Logical Relationships:\n{invalid_logic}")
    return invalid_logic
```

```python
def check_skip_logic(data):
    invalid_skip = data[
        (data['AI_Usage_Level'] == 'Not used') & 
        (data['AI_Applications'] != 'No response')
    ]
    print(f"\nRows Violating Skip Logic:\n{invalid_skip}")
    return invalid_skip
```

```python
def check_timestamps(data, timestamp_column):
    data[timestamp_column] = pd.to_datetime(data[timestamp_column], errors='coerce')
    invalid_dates = data[data[timestamp_column] > pd.Timestamp.now()]
    print(f"\nInvalid Timestamps:\n{invalid_dates}")
    return invalid_dates
```

```python
def generate_validation_report(data):
    print("\n--- Validation Report ---")
    
    # Missing Values
    missing_values = check_missing_values(data)
    
    # Duplicates
    duplicates = check_duplicates(data)
    
    # Consistency Check for Example Column
    valid_ai_usage_levels = ['Not used', 'Limited use', 'Moderate use', 'Extensive use']
    invalid_values = check_categorical_consistency(data, 'AI_Usage_Level', valid_ai_usage_levels)
    
    # Logical Relationships
    logical_issues = check_logical_relationships(data)
    
    # Skip Logic
    skip_logic_issues = check_skip_logic(data)
    
    # Timestamps
    invalid_timestamps = check_timestamps(data, 'Timestamp')
    
    # Summary
    report = {
        "Missing Values": missing_values.sum(),
        "Duplicate Rows": duplicates,
        "Invalid Categorical Values": invalid_values,
        "Logical Issues": len(logical_issues),
        "Skip Logic Issues": len(skip_logic_issues),
        "Invalid Timestamps": len(invalid_timestamps)
    }
    print("\nSummary of Validation Issues:\n", report)
    return report
```

```python
# Run the validation report
validation_report = generate_validation_report(survey_data)

```
